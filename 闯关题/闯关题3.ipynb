```python
# 使用 `PreliZ` 定义 Beta 分布，参数 `alpha = 2`, `beta = 5`，并生成符合分布的样本。
# np.random.seed(123)
dist = pz.Beta(2, 5)
dist.rvs(1000, random_state=123)
```

## STEP3：概率分布的统计量

定义 Gamma 分布，参数 `alpha = 2`, `beta = 1`，计算其统计量。
1. 问题q1：计算 Gamma 分布的均值和中位数。
     - 提示，可以使用 `.median` 方法计算中位数，或者先通过 `.rv` 方法生成样本，在通过 `np.median` 计算中位数。
2. 问题q2：计算 Gamma 分布的最高密度区间(highest density interval)。
3. 问题q3：通过 `.summary` 显示 Gamma 分布的常见统计量。
4. 问题q4：计算 Gamma 分布在 [0.1,0.5,0.9]的百分位数。提示：`.ppf`。
5. 问题q5：计算数据值 `x=3` 在 Gamma 分布下的对数似然值。提示：`.logpdf`。

```python
dist = pz.Gamma(2, 1)

# 问题q1：问题q1：计算 Gamma 分布的均值和中位数。

# 计算均值
mean = gamma_dist.mean()
print(f'Mean: {mean}')

# 计算中位数
median = gamma_dist.median()
print(f'Median: {median}')
```

```python
# 问题q2：问题q2：计算 Gamma 分布的最高密度区间(highest density interval)。

# 计算最高密度区间
hdi = gamma_dist.hdi()
print(f'Highest Density Interval: {hdi}')
```

```python
# 问题q3：问题q3：通过 `.summary` 显示 Gamma 分布的常见统计量。

# 显示常见统计量
summary = gamma_dist.summary()
print(summary)
```

```python
# 问题q4：计算 Gamma 分布在 [0.1,0.5,0.9]的百分位数。

# 计算百分位数
percentiles = [0.1, 0.5, 0.9]
percentile_values = gamma_dist.ppf(percentiles)
print(f'Percentiles at {percentiles}: {percentile_values}')
```

```python
# 问题q5：计算数据值 `x=3` 在 Gamma 分布下的对数似然值。

# 计算对数似然值
x = 3
log_likelihood = gamma_dist.logpdf(x)
print(f'Log-Likelihood for x={x}: {log_likelihood}')
```

## STEP4：通过数据和统计量计算概率分布

1. 问题q1：通过最大熵方法 `pz.maxent` 计算区间在[0.3, 0.8]之间，概率为0.6的正态分布的均值和方差。
      - 提示，pz.maxent 输入的第一个参数为分布，第二、三个参数为区间的上下界，第四个参数为该区间的概率占比。
2. 问题2：通过 `pz.quartile` 方法计算四分位值为 [1,3,5] 下正态分布的均值和方差。
      - 提示，pz.quartile 输入的第一个参数为分布，第二至五个参数为四分位值。

```python
# 问题q1：通过最大熵方法 `pz.maxent` 计算区间在[0.3, 0.8]之间，概率为0.6的正态分布的均值和方差。
# dist = pz.Normal()
# pz.maxent(dist, ...)

# 定义正态分布
normal_dist = pz.Normal()

# 计算在区间 [0.3, 0.8] 之间概率为 0.6 的正态分布的均值和方差
mean, variance = pz.maxent(normal_dist, 0.3, 0.8, 0.6)
print(f'Mean: {mean}, Variance: {variance}')
```

```python
# 问题2：通过 `pz.quartile` 方法计算四分位值为 [1,3,5] 下正态分布的均值和方差。
# 计算四分位值为 [1, 3, 5] 下正态分布的均值和方差

# 定义正态分布
normal_dist = pz.Normal()

# 计算四分位值为 [1, 3, 5] 下正态分布的均值和方差
quartiles = [1, 3, 5]
mean_quartile, variance_quartile = pz.quartile(normal_dist, *quartiles)
print(f'Mean: {mean_quartile}, Variance: {variance_quartile}')
```
