\end{align*}
$$

## 贝叶斯因子（Bayes Factors， BFs）


贝叶斯因子（Bayes Factors，BFs）是用来量化一个模型相对于另一个模型的相对证据的指标。
* 贝叶斯因子通过后验概率比/先验概率比，反应数据对假设的支持程度。
* 在贝叶斯推断框架下，贝叶斯因子起到与经典频率主义框架中 p 值类似的作用，都是用于假设检验的工具。

贝叶斯框架中的假设检验与经典的显著性检验有所不同。
* 在经典统计中，p 值用于评估在原假设为真的情况下，观察到的数据有多不可能；
* 而在贝叶斯模型选择框架中，贝叶斯因子则用于比较不同模型之间的证据强度，每个模型对应一个具体的假设。

**贝叶斯定理与贝叶斯因子**

根据贝叶斯定理，我们可以根据观测到的数据更新某个模型的先验概率（$P(M)$）为后验概率（$P(M|D)$）。更新的过程包括考虑在该模型下观察到数据的可能性，即似然（$P(D|M)$）。贝叶斯因子可以通过比较两个模型的后验概率比来得到：

$$  \frac{P(M_1|D)}{P(M_2|D)} = \frac{P(D|M_1)}{P(D|M_2)} \times \frac{P(M_1)}{P(M_2)}
$$

在这个公式中，$\frac{P(D|M_1)}{P(D|M_2)}$ 就是贝叶斯因子，它表示在两个模型下，数据的似然比；而 $\frac{P(M_1)}{P(M_2)}$ 是先验比，即两个模型的先验概率比。贝叶斯因子可以解释为，基于观察数据，如何更新先验关于两个模型的信念，从而得到后验概率。

**贝叶斯因子的计算**
贝叶斯因子可以通过两种方式来计算：
1. **通过数据的相对可能性**：即通过观察数据在两个模型下的似然比来计算。具体公式为：

$$  BF_{12} = \frac{P(D|M_1)}{P(D|M_2)}
$$

2. **通过先验信念的转变**：贝叶斯因子还可以通过计算先验信念的变化来得到，即通过计算从先验比到后验比的转变：

$$  BF_{12} = \frac{\text{Posterior Odds}}{\text{Prior Odds}}
$$



**结果解读**

在假设检验的框架中，贝叶斯因子在贝叶斯统计框架中的作用类似于经典频率学派框架中的$p$值。

在显著性检验中，$p$值用于评估在零假设为真的情况下观测数据出现的概率，而在贝叶斯模型选择框架中，贝叶斯因子则用于评估不同模型的证据，每个模型对应一个特定的假设。

贝叶斯因子与传统的显著性检验（如 p 值）不同，后者仅关心假设是否被拒绝，而贝叶斯因子则提供了一个量化的指标，表明数据对某个假设或模型的支持强度。因此，贝叶斯因子不仅可以用于比较两个假设的支持，还可以提供更多的信息，帮助研究者在假设检验过程中做出更合理的决策。

### 基于自我匹配范式的简单线性回归模型 

我们还是以第八节课的随机点运动范式的简单线性模型为例进行演示，我们需要回答 "贝叶斯因子能用来回答什么问题？"

**简单线性回归模型的数学表达式：**  


$$  
\begin{align*}  
\beta_0   &\sim N\left(m_0, s_0^2 \right)  \\  
\beta_1   &\sim N\left(m_1, s_1^2 \right)  \\  
\sigma    &\sim \text{Exp}(\lambda)        \\  
&\Downarrow \\  
\mu_i &= \beta_0 + \beta_1 X_i             \\  
&\Downarrow \\  
Y_i | \beta_0, \beta_1, \sigma &\sim N\left(\mu_i, \sigma^2\right). \\  
\end{align*}  
$$  

![Image Name](https://cdn.kesci.com/upload/smkijcu8co.png?imageView2/0/w/700)  

贝叶斯因子能够回答一些关于假设、模型和数据之间关系的问题，以下是贝叶斯因子常见的三种应用及可能的假设示例：

假设我们想检验自变量 $X$ 是否对因变量 $Y$ 有显著影响，即检验 $\beta_1 = 0$（无效假设）与 $\beta_1 \neq 0$（有效假设）之间的关系。我们可以使用贝叶斯因子来比较这两个模型，计算贝叶斯因子 $BF_{01}$，来量化零假设和备择假设之间的证据差异。

如果贝叶斯因子值较大（例如 $BF_{01} = 10$），则说明数据支持 $\beta_1 \neq 0$ 的假设，即自变量 $X$ 对因变量 $Y$ 的影响显著。反之，如果贝叶斯因子较小（例如 $BF_{01} = 0.1$），则表明数据支持 $\beta_1 = 0$，即认为 $X$ 对 $Y$ 的影响不显著。


1. **与点零假设（0）比较（Testing against the point-null）**  
   贝叶斯因子可以用来检验某个参数是否等于零，即检验点零假设。例如，在简单线性回归中，我们可能提出假设 $H_0: \beta_1 = 0$，认为自变量 $X$ 对因变量 $Y$ 没有影响。通过贝叶斯因子，我们可以量化数据支持零假设与备择假设（$H_1: \beta_1 \neq 0$）之间的证据差异。
2. **与零区域比较（Testing against a null-region）**  
   有时，我们可能会设定一个“零区域”，即一个参数值的范围，该范围内的值被认为与零没有显著差异。例如，假设我们设定一个区间 $\beta_1 \in [-1, 1]$，认为回归系数 $\beta_1$ 落在这个区间内是可以接受的无效效应。通过贝叶斯因子，我们可以计算数据支持回归系数在零区域内（$\beta_1 \in [-1, 1]$）或零区域外（$\beta_1 \notin [-1, 1]$）的证据强度。
3. **方向性假设（Directional hypotheses）**  
   贝叶斯因子也可以用于检验某个参数是否有特定的方向性，即正向或负向。例如，我们可以提出假设 $H_1: \beta_1 > 0$，认为自变量 $X$ 对因变量 $Y$ 有正向影响。贝叶斯因子将帮助我们评估数据是否支持这种方向性的假设，而不仅仅是支持或拒绝零假设。


## 基于点估计的检验方法 (Testing against the point-null, 与点零假设（0）比较)

🤔在本节课的线性模型中，参数 $\beta_1$ 代表了两个条件下被试反应速度的差异。我们脑中可能自然生成这样一个问题：

> 两个条件下的反应速度是否相同？ 即 $\beta_1$ 是否等于零？

当我们测试参数是否等于零时（如检验回归系数 $\beta_1 = 0$），这就是点零假设的检验。贝叶斯因子可以帮助我们衡量数据对点零假设与非零假设的支持程度。在回归模型中，我们可以设定一个原假设 $H_0: \beta_1 = 0$ 和备择假设 $H_1: \beta_1 \neq 0$，然后使用贝叶斯因子来计算零假设和备择假设的相对证据。

$$
BF_{01} = \frac{P(Data|\beta_1 \neq 0)}{P(Data|\beta_1 = 0)}
$$

我们还是以编号为"201"的被试数据为例进行演示：

```python
# 导入 pymc 模型包，和 arviz 等分析工具 
import pymc as pm
import arviz as az
import scipy.stats as st
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import pandas as pd

# 忽略不必要的警告
import warnings
warnings.filterwarnings("ignore")

# 通过 pd.read_csv 加载数据 Kolvoort_2020_HBM_Exp1_Clean.csv
try:
  df_raw = pd.read_csv('/home/mw/input/bayes3797/Kolvoort_2020_HBM_Exp1_Clean.csv')
except:
  df_raw = pd.read_csv('data/Kolvoort_2020_HBM_Exp1_Clean.csv')

# 筛选出被试"201"，匹配类型为"Matching"的数据
df_raw["Subject"] = df_raw["Subject"].astype(str)
df = df_raw[(df_raw["Subject"] == "201") & (df_raw["Matching"] == "Matching")]

# 选择需要的两列
df = df[["Label", "RT_sec"]]

# 重新编码标签（Label）
df["Label"] = df["Label"].map({1: 0, 2: 1, 3: 1})

# #设置索引
df["index"] = range(len(df))
df = df.set_index("index")

# 显示部分数据
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 计算每个Label条件下的均值
mean_values = df.groupby('Label')['RT_sec'].mean()

# 使用 seaborn 绘制箱线图，展示 Matching 条件下反应时间的分布情况
plt.figure(figsize=(5, 3.2))
sns.boxplot(x="Label", y="RT_sec", data=df)

plt.plot(mean_values.index, mean_values.values, marker='o', color='r', linestyle='-', linewidth=2)

# 移除上面和右边的边框
sns.despine()

# 添加标签
plt.xlabel("Label Condition (0=self, 1=other)")
plt.ylabel("Reaction Time (sec)")
plt.show()
```

对数据进行可视化后，我们可以看出，在other条件下的反应时间较长。

**那么我们想知道，在self和other的条件下这个反应时间的差异到底有多大**，我们就可以通过贝叶斯因子的方式对这个问题进行解释。

在该分析中，我们关注的是两个条件之间的差异，因此在模型中的三个参数 beta_0、beta_1和sigma中，

我们仅重点关注**beta_1**，因为它反映了self和other条件之间的反应时间差异。

传统的零假设显著性检验（Null hypothesis significance test, NHST）的框架之下，*t*检验只提供了一个二分的结果：拒绝或者无法拒绝$H_0$。 

贝叶斯因子方法可以提供一种更直观和渐进的方式来评估数据支持原假设（无效假设，$H_0$）还是备择假设($H_1$)。 

这意味着我们不仅依赖数据，还依赖于我们的先验假设，这与传统的频率学派方法不同。


**回顾：建立贝叶斯模型**


根据我们的模型：

$$
Y_i \mid \beta_0, \beta_1, \sigma \sim \mathcal{N}(\mu_i, \sigma^2)
$$

**先验（prior）**  

* > $\beta_{0}   \sim N\left(5, 2^2 \right)$  
  * > 模型的截距项服从均值为 5，标准差为 2 的正态分布。  
* > $\beta_1   \sim N\left(0, 1^2 \right)$  
  * > 模型的斜率项，服从均值为 0，标准差为 1 的正态分布。  
* > $\sigma   \sim \text{Exp}(0.3)$  
  * > 代表误差项的标准差，服从参数为 0.3 的指数分布。  

**似然（likelihood）**  

* > $\mu_i = \beta_0 + \beta_1X_i$  
* > $Y_i {\sim} N\left(\mu_i, \sigma^2\right)$  


<div style="padding-bottom: 30px;"></div>

```python
# 设置随机种子以确保结果可复现
np.random.seed(123)

with pm.Model() as linear_model:

    # 定义先验分布参数
    beta_0 = pm.Normal("beta_0", mu=5, sigma=2)        
    beta_1 = pm.Normal("beta_1", mu=0, sigma=1)      
    sigma = pm.Exponential("sigma", 3)                    

    # 定义自变量 x
    x = pm.MutableData("x", df['Label'])         

    # 定义 mu，将自变量与先验结合
    mu = beta_0 + beta_1 * x

    # 定义似然：预测值y符合N(mu, sigma)分布
    likelihood = pm.Normal("y_est", mu=mu, sigma=sigma, observed=df['RT_sec'])  
```

```python
with linear_model:
    trace = pm.sample(draws=5000,                   # 使用mcmc方法进行采样，draws为采样次数
                      tune=1000,                    # tune为调整采样策略的次数，可以决定这些结果是否要被保留
                      chains=4,                     # 链数
                      discard_tuned_samples=True,  # tune的结果将在采样结束后被丢弃
                      random_seed=84735)
```

```python
trace.posterior['beta_1']
```

### Arviz 实现计算贝叶斯因子 (BF)

对于点估计的贝叶斯因子检验，常用的计算方法是 **Savage-Dickey 比率**（Savage-Dickey ratio），特别适用于检验零假设是否等于某个具体值。
* Savage-Dickey 方法可以通过比较该参数在后验分布和先验分布中的密度来计算贝叶斯因子。

在 Python 的 Arviz 库中，`arviz.plot_bf` 函数提供了直接计算贝叶斯因子的方法，并且可以进行图形化的展示。

```python
beta_1_posterior = trace.posterior['beta_1'].stack(flat_dim=("chain", "draw")).values
```

```python
idata = az.from_dict(
    posterior={"beta_1": trace.posterior['beta_1'].values},
    prior={"beta_1": np.random.normal(0, 1, 5000)}  # 假设先验分布
)

az.plot_bf(idata, var_name="beta_1", ref_val=0)
plt.xlim(-0.5, 0.5) 
sns.despine()
```

通过上图我们可以发现：

1. BF_10 = 5.80 表示数据支持假设 $H_1$ 的程度是支持假设 $H_0$ 的 5.80 倍。
2. BF_01 = 0.17 表示数据支持假设 $H_0$ 的程度是支持假设 $H_1$ 的 0.17 倍。


为了凸显先验对于计算BF的影响，我们给定一个虚假的先验分布 $prior \sim N(0,0.1)$。

```python
idata = az.from_dict(
    posterior={"beta_1": trace.posterior['beta_1'].values},
    prior={"beta_1": np.random.normal(0, 0.1, 5000)}  # 假设先验分布
)

az.plot_bf(idata, var_name="beta_1", ref_val=0)
plt.xlim(-0.5, 0.5) 
sns.despine()
```

### 与 JASP 的对比

另一种常见的计算贝叶斯因子（BF）的方法来源于 JASP（Jeffreys's Amazing Statistics Program）等现代统计工具。
* JASP 是一个基于 R 语言的开源软件，以其易用性和友好的图形界面广受欢迎，主要用于心理学、社会学等领域的统计分析。
* JASP 使用贝叶斯模型来计算贝叶斯因子，包括模型比较和桥采样法（Bridge Sampling）等技术，能够通过计算后验和先验分布的比值，评估不同模型或假设的相对支持程度。与 Savage-Dickey 方法不同，桥采样不局限于单参数或简单假设，因此更适合模型比较和复杂假设检验。

**Python 版本的 JASP 实现**

在 Python 中，虽然没有官方的 JASP API，但我们可以使用 `arviz.compare` 实现类似 JASP 中模型比较的方法来计算贝叶斯因子。
* 需要注意的是，要计算贝叶斯因子，需要获得模型的边际似然（Marginal Likelihood），即模型中所有参数的联合似然。
* PyMC 提供了 `pm.compute_log_likelihood` 方法，可以计算模型中所有参数的联合似然。

> 在`arviz.compare()`的输出中，没有直接提供贝叶斯因子（BF）的值。贝叶斯因子是用来比较模型的似然（或后验概率）的一种度量，可以通过**权重(weight)**来间接理解模型的相对优劣，但它不是直接的贝叶斯因子。
> $$\text{BF}_{10} = \frac{\text{weight of model 1}}{\text{weight of model 2}}$$

注意：由于这里的方法基于模型比较，它更适合模型比较和复杂假设检验，因此和 Savage-Dickey 方法不同。
但可以看到两者的结果是接近的：$BF_{10}(SD方法)=5.8$；$BF_{10}(模型比较)=9.5$。

```python
# 定义零假设模型（仅包含截距的模型）
with pm.Model() as model_H0:

    beta_0 = pm.Normal("beta_0", mu=5, sigma=2)        
    sigma = pm.Exponential("sigma", 3)                    
    mu = beta_0
    likelihood = pm.Normal("y_est", mu=mu, sigma=sigma, observed=df['RT_sec'])  
    trace_H0 = pm.sample(draws=5000, tune=1000, chains=4,discard_tuned_samples=True, random_seed=84735)
    pm.compute_log_likelihood(trace_H0)

# 定义备择假设模型（包含截距和斜率的模型）
with pm.Model() as model_H1:
    
    beta_0 = pm.Normal("beta_0", mu=5, sigma=2)        
    beta_1 = pm.Normal("beta_1", mu=0, sigma=1)      
    sigma = pm.Exponential("sigma", 3)                    
    x = pm.MutableData("x", df['Label'])         
    mu = beta_0 + beta_1 * x
    likelihood = pm.Normal("y_est", mu=mu, sigma=sigma, observed=df['RT_sec'])  
    trace_H1 = pm.sample(draws=5000, tune=1000, chains=4,discard_tuned_samples=True, random_seed=84735)
    pm.compute_log_likelihood(trace_H1)
```

```python
# 计算贝叶斯因子
model_compare = az.compare({"Label": trace_H0, "Null model": trace_H1}, method='BB-pseudo-BMA')
weight_H0 = model_compare.loc["Label", "weight"]
weight_H1 = model_compare.loc["Null model", "weight"]
BF_10 = weight_H1 / weight_H0
print(f"贝叶斯因子 (BF_10): {BF_10}")
```

**点估计的局限性**

虽然点估计方法为我们提供了参数的一个最佳估计值，但它没有反映参数不确定性的范围。

-  一个特定的点估计不容易准确,因为它只是许许多多点估计中的一个。
- 点估计没有提供与所估计参数有多相近的点估计值的概率信息。
- 区间估计比点估计提供更多的有关总体特征的信息。

为了解决这一问题，我们还可以基于区间估计的方法来理解参数的取值范围以及数据所支持的参数区间。

区间估计可以通过后验分布来实现，并进一步计算最高密度区间（Highest Density Interval, HDI）或置信区间。


## 针对区间的检验方法 (Testing against a null-region, 与零区域比较)

**回顾 ROPE**

上节课我们学习了使用 HDI + ROPE 进行检验。我们假设了 ($\beta_1$) 的值在 $[-0.05, 0.05]$ 范围内可以视为实用等效，  

即如果$\beta_1$落在这个范围内，说明 self 和 other 条件之间的反应时间差异可以忽略不计，从而在实践上认为两者无显著差异。  

**ROPE 对于贝叶斯因子的意义**

当我们进行假设检验时，有时会设定一个“零区域（null region）”，即参数值的一个范围，来定义在实际意义上可以视为“无效”的效果范围。

通过设定零区域，我们可以计算先验概率和后验概率，进而计算贝叶斯因子来量化数据对假设的支持强度。

我们可以计算 "label" 条件下反应时间差异落在零假设区间之外的先验概率，以及落在零假设区间之内的先验概率，从而得到我们的先验概率比（prior odds）：


$$
BF_{01} = \frac{P(Data| \beta_1 \notin [-0.05, 0.05])}{P(Data| \beta_1 \in [-0.05, 0.05])}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 定义先验分布（beta_1）参数
mean = 0  
std_dev = 1 

# 定义零假设区间
null_region = (-0.05, 0.05)

# 定义x轴的范围
x = np.linspace(-4, 4, 1000)

# 计算正态分布的密度
y = norm.pdf(x, mean, std_dev)

# 创建图形
plt.figure(figsize=(8, 5))

# 绘制密度曲线
plt.plot(x, y, color='black')

# 填充零假设区间的颜色
plt.fill_between(x, y, where=(x >= null_region[0]) & (x <= null_region[1]), color='blue', alpha=0.6, label='Null')

# 填充备择假设区间的颜色
plt.fill_between(x, y, where=(x < null_region[0]) | (x > null_region[1]), color='yellow', alpha=0.6, label='Alternative')

# 添加标签和图例
plt.xlabel("Label effect")
plt.ylabel("Density")
plt.legend(loc='upper right', title="Prior regions")
plt.title("Prior Distribution with Null Region")

sns.despine()
plt.show()
```

```python
import scipy.stats as stats

# 定义beta_1先验分布的参数
mu = 0
sigma = 1

# 计算区间 [-0.05, 0.05] 内的概率
p_in = stats.norm.cdf(0.05, mu, sigma) - stats.norm.cdf(-0.05, mu, sigma)

# 计算区间外的概率
p_out = 1 - p_in

# 计算先验概率比
prior_odds = p_out / p_in

print(f"先验概率比（prior odds）: {prior_odds:.4f}")
```

此时，我们可以通过计算后验概率比（Posterior Odds），评估数据更新后参数值在零假设区间内和区间外的相对支持程度：

$$ BF_{10} = \frac{P(\beta_1 \notin [-0.05, 0.05] \mid \text{Data})}{P(\beta_1 \in [-0.05, 0.05] \mid \text{Data})} $$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 定义零假设区间
null_region = (-0.05, 0.05)

# 定义x轴范围
x = np.linspace(-0.4, 0.4, 1000)

# 计算先验和后验的密度
prior_density = norm.pdf(x, 0, 1)
# COMMENT: 这里用标准差计算可能不太准确，后续看能不能优化为直接根据采样进行计算。
posterior_density = norm.pdf(x, 0.149, 0.047)

# 计算后验概率比（Posterior Odds）
# 使用累积分布函数（CDF）计算落在 null_region 内和外的概率
p_in_null_region = norm.cdf(null_region[1], 0.149, 0.047) - norm.cdf(null_region[0], 0.149, 0.047)
p_outside_null_region = 1 - p_in_null_region
posterior_odds = p_outside_null_region / p_in_null_region

# 绘制图形
plt.figure(figsize=(10, 8))

# 绘制后验分布的密度曲线
plt.plot(x, posterior_density, color='black')

# 填充零假设区间的颜色
plt.fill_between(x, posterior_density, where=(x >= null_region[0]) & (x <= null_region[1]), color='blue', alpha=0.6, label='Null')

# 填充备择假设区间的颜色
plt.fill_between(x, posterior_density, where=(x < null_region[0]) | (x > null_region[1]), color='yellow', alpha=0.6, label='Alternative')

# 添加概率比文本
plt.text(0.4, 0.35, f"Posterior Odds: {posterior_odds:.2f}", fontsize=12, color='black')

# 设置标签和标题
plt.xlabel("Label effect")
plt.ylabel("Density")
plt.legend(loc='upper right', title="Posterior regions")
plt.title("Posterior Distribution with Null Region")

sns.despine()
plt.show()
```
